```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**Matplotlib Basics**

```python
x = np.linspace(0, 5, 11)
print(x)
y = x ** 2
print(y)
```

```python
plt.plot(x, y,'g')
```

```python
plt.xlabel('X Label')#any thing can write
plt.ylabel('Y Label')
plt.title('Title')
plt.plot(x, y)
```

**Subplots in Matplotlib**

```python
plt.subplot(2,2,1) #subplot -> divided in rows and column
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
```

**Object Oriented way**

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.

figsize is a tuple of the width and height of the figure in inches
dpi is the dots-per-inch (pixel per inch).

```python
fig = plt.figure(figsize=(10,6),dpi=300) #dpi-> clearence figsize->size of canvas
axis1 =fig.add_axes([0.1,0,0.8,0.8])#canvas er upor axis
axis1.plot(x,y)
axis2 = fig.add_axes([0.2,0.3,0.4,0.3]) #control over axis
axis2.plot(y,x)
```
