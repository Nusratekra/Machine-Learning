# ***Character Class Set:[]***


*   [aeiou]: match any vowel
*   [0 -9]: match any digit
*   [a - z]: match any lower case
*    [A - Z]: match any lower caseList item



```python
pattern= "[aeiou]"
result= re.search(pattern, "Nusrat Ahmmed Ekra")
if result:
  print("Vowel Match")
else:
  print("Vowel Not Match")
```

```python
pattern= "[aeiou]"
result= re.search(pattern, "xyz")
if result:
  print("Vowel Match")
else:
  print("Vowel Not Match")
```

```python
pattern= "[0-9]"
result= re.search(pattern, "12345")
if result:
  print("digit Match")
else:
  print("digit Not Match")
```

```python
pattern= "[0-9]"
result= re.search(pattern, "Ekra12345")
if result:
  print("digit Match")
else:
  print("digit Not Match")
```

# ***Kleenee closure and star***
*   "*": 0 or more
*   "+": 1 or more
*   new(s)* : new,news,newss,newsss,newsss
*   new(s)+ : news,newss,newsss,newsss
*   Star or plus can be use in middle also





```python
pattern= "new(s)*"
result= re.match(pattern, "new")
if result:
  print("Match")
else:
  print("Not Match")
```

```python
pattern= "new(s)*"
result= re.match(pattern, "newsssssss")
if result:
  print("Match")
else:
  print("Not Match")
```

```python
pattern= "new(s)+"
result= re.match(pattern, "new")
if result:
  print("Match")
else:
  print("Not Match")
```

```python
pattern= "new(s)+"
result= re.match(pattern, "newss")
if result:
  print("Match")
else:
  print("Not Match")
```

```python
pattern= "FRI(E)*nd"
result= re.match(pattern, "FRIEnd")
if result:
  print("Match")
else:
  print("Not Match")
```

```python
pattern= "FRI(E)*nd"
result= re.match(pattern, "FRInd")
if result:
  print("Match")
else:
  print("Not Match")
```

```python
pattern= "FRI(E)+nd"
result= re.match(pattern, "FRInd")
if result:
  print("Match")
else:
  print("Not Match")
```

```python
pattern= "FRI(E)+nd"
result= re.match(pattern, "FRIEnd")
if result:
  print("Match")
else:
  print("Not Match")
```

# ***Pipe symbol(|)*** -->
*   Works like logical OR
*   (Chooseing sub string from total string)

```python
pattern= "Go(es|ing)"
text= "Goes"
if re.match(pattern, text):
  print("Match")
else:
  print("Not Match")
```

```python
pattern= "Go(es|ing)"
text= "Going"
if re.match(pattern, text):
  print("Match")
else:
  print("Not Match")
```

```python
pattern= "Go(es|ing)"
text= "Goif"
if re.match(pattern, text):
  print("Match")
else:
  print("Not Match")
```
