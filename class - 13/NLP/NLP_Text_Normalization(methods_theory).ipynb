# ***Text Normalization(converting into a standard form)***
*   word boundery detection
*   Separated word from each other
*   Example 1: Bangladesh, New York, cats and dog
*   Example 2: #nlp, @dipubanik, emojies,:)

***Methods of Text Normalization:***
*   Tokenization:
    *   input: Bangladesh is beautiful
    *   output:[Bangladesh, is, beautiful]
*   Lemmatization: (Finding the same root)
    *   input : [sings, sung, sang] ,[connection, connected, connecting]
    *   output : [sing] , [connect]
*   Stemming: Porter stemmer(main word find out)
    *   rules1: atinal -> ate, ator -> ate
    *   example1 : Relational -> relate , operato -> operate
    *   rules2: ing -> null, s -> null
    *   example2: going -> go, walking -> walk, cats -> cat
*   Stemming challange: need to handel some exeptional case
    *   input: king,nothing, news
    *   output: k,noth,new
*   Stop words: Remove the common word that are not important
    *    Words: am,is,are,an,the
    *   List item
        *   Example:
              *   Input: KUET is a beautiful campus
              *    Output: KUET beautiful campus
*   POS(parts of speech) Tagging: Identify the grammar of each word
      *   Input : KUET is a beautiful campus.
      *   Output : [KUET= Noun, is =Verb, a =Article, beautiful=Adjective, campus=Noun]
