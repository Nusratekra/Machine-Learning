# ***Sentence Tokenization***

```python
from nltk.tokenize import sent_tokenize
```

```python
text="The Department of Computer Science and Engineering at KUET began its academic journey on September 26, 1999, with a commitment to delivering high-quality education and conducting cutting-edge research in various fields of computer science and engineering. The department offers undergraduate (B.Sc.) as well as postgraduate (M.Sc. and Ph.D.) programs. Every year, only the most talented and top-ranked students are admitted through a highly competitive admission test. Currently, the department enrolls around 160 students annually, including a fixed intake of 120 in the undergraduate program. Well-equipped laboratories and a dedicated System Development Centre provide students with strong practical exposure alongside their theoretical learning."
text
```

```python
sent_tokenize(text)
```

# ***Word Tokenization***

```python
from nltk.tokenize import word_tokenize
wt = word_tokenize(text)
wt
```

# ***White Space***

```python
from nltk.tokenize import WhitespaceTokenizer
WhitespaceTokenizer().tokenize(text)
```

# ***Word Punctuation tokenization***

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(text)
```

# ***Frequency Distribution(Statical analysis of each word)***

```python
from nltk.probability import FreqDist
fdist = FreqDist(wt)
print(fdist)
```

```python
fdist.most_common() #count
```

```python
fdist.most_common(5)
```

# ***Plot Frequency Distribution***

```python
import matplotlib.pyplot as plt
fdist.plot(35,cumulative=False)
plt.show()
```

```python
fdist.plot(15,cumulative=False)
plt.show()
```

```python
fdist.plot(15,cumulative=True) #if True than in count format
plt.show()
```
