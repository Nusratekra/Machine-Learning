### *The minimum number of editing operations need to be transform one string to another*

# ***Dynamic Edit Distance***

```python
import numpy as np
```

```python
text1 = 'deal'
text2 = 'leda'
```

# ***Initialize the algorithm***

```python
dist = np.zeros((len(text1)+1,len(text2)+1))
dist[0,:] = np.arange(len(text2)+1)
dist[:,0] = np.arange(len(text1)+1)
dist
```

# ***Dynamic Programming to calculate minimum edit distance***

```python
for i in range(1,len(text1)+1):
  for j in range(1,len(text2)+1):
    if text1[i-1] == text2[j-1]:
      dist[i,j] = dist[i-1,j-1]
    else:
      dist[i,j] = min(dist[i-1,j]+1,dist[i,j-1]+1,dist[i-1,j-1]+2)
dist
```
