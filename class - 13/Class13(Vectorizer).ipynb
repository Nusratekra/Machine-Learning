```python
import nltk
nltk.download('punkt')
from nltk.tokenize import word_tokenize
```

```python
text_vsc = [word_tokenize(test) for test in df['test']]
print(text_vsc)
```

```python
text = []
for test in df['test']:
    text.append(nltk.word_tokenize(test))
text
```

```python
model = Word2Vec(text,min_count=1) #min_count=1 => eikhane 1 bar .same type value kobar hole hount korbe,by default its 5
```

```python
model
```
